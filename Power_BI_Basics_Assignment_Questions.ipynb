# Question 1 : What is Power BI and why is it used in businesses?

# Ans-->


**Power BI** is a **Business Intelligence (BI) and data visualization tool developed by Microsoft**. It is used to **connect, analyze, and visualize data** from different sources and turn it into **interactive reports and dashboards** that help businesses make better decisions.

---

## 1. What is Power BI?

![Image](https://learn.microsoft.com/en-us/power-bi/create-reports/media/service-dashboards/power-bi-dashboard2.png)

![Image](https://cdn.prod.website-files.com/68e9e85a5b12150471bd3ee4/68ff3a0e32efe2e47fae2115_supply-chain-dashboard.svg)

![Image](https://its.weill.cornell.edu/sites/default/files/styles/panopoly_image_original/public/dashboard-mockup.png?itok=FJOBxrkN)

![Image](https://blog.bismart.com/hubfs/Imported_Blog_Media/data%20visualization%20with%20power%20bi-Sep-26-2023-09-10-49-2729-AM.jpg)

**Power BI** allows users to:

* Connect data from **multiple sources** (Excel, SQL databases, cloud services, etc.)
* **Clean and transform data** using Power Query
* Create **interactive charts, graphs, and dashboards**
* Share insights with teams through **online reports**

It converts **raw data into meaningful visual insights**.

---

## 2. Why Businesses Use Power BI

Businesses use Power BI for several important reasons:

### 1️⃣ Data Visualization

Power BI turns complex data into **easy-to-understand visuals** like:

* Bar charts
* Pie charts
* Maps
* KPI indicators

This helps managers understand trends quickly.

---

### 2️⃣ Better Decision Making

Power BI provides **real-time insights**, helping companies make **data-driven decisions instead of guessing**.

Example:
A company can analyze **sales data** to see which product sells the most.

---

### 3️⃣ Connect Multiple Data Sources

Power BI can connect to many sources such as:

* Excel
* SQL Server
* MySQL
* Cloud services
* APIs

For someone like you learning **data analytics**, this is very useful because analysts often combine data from multiple systems.

---

### 4️⃣ Interactive Dashboards

Users can **filter, drill down, and explore data interactively**.

Example:
A manager can click on a **region** in the dashboard and see **sales for that region only**.

---

### 5️⃣ Easy Report Sharing

Reports can be shared through **Power BI Service (cloud)** so teams can access them anywhere.

---

## 3. Simple Real-World Example

Suppose a company has **sales data of 10,000 orders**.

Using Power BI, a data analyst can create dashboards showing:

* Total sales by **month**
* Sales by **city**
* Top **5 products**
* Profit by **category**

This helps management quickly understand **business performance**.

# Question 4 : List any four data cleaning tasks that can be performed in Power Query.

# Ans -->

In **Power Query** (a data transformation tool in **Power BI** by Microsoft), data analysts can clean and prepare raw data before analysis.

### Four Data Cleaning Tasks in Power Query

1️⃣ **Remove Duplicates**
Power Query can remove duplicate records from a dataset so that each entry appears only once.
*Example:* If the same customer appears multiple times, duplicates can be removed to maintain accurate data.

2️⃣ **Handle Missing Values**
Missing or blank values can be replaced, filled, or removed.
*Example:* Replace blank values in a column with **0**, **average value**, or **“Unknown”**.

3️⃣ **Change Data Types**
Data types can be corrected or modified.
*Example:* Converting a column from **text to number**, or **text to date** so calculations can be performed properly.

4️⃣ **Filter Rows**
Unnecessary rows can be filtered out based on conditions.
*Example:* Removing records where **sales amount = 0** or selecting data only for a specific year.
